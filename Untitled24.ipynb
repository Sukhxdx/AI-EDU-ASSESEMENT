```python
import streamlit as st
from modules.quiz_generator import QuizGenerator
from modules.readability import ReadabilityAnalyzer
from modules.glossary import GlossaryBuilder
from modules.db import Database

st.set_page_config(page_title='AI Edu Assessment', layout='wide')

st.sidebar.title('AI Edu Assessment')
page = st.sidebar.radio('Go to', ['Home', 'Analyze Text', 'Quizzes', 'Settings'])

# Initialize shared components
qg = QuizGenerator()
ra = ReadabilityAnalyzer()
gb = GlossaryBuilder()
db = Database('data/projects.db')

# ---- PAGE ROUTING ----

if page == 'Home':
    st.title('🎓 AI-Powered Educational Assessment Platform')
    st.markdown('**Advancing SDG 4 (Quality Education), SDG 9 (Innovation & Infrastructure), '
                'and SDG 10 (Reduced Inequalities)**')
    st.markdown('''
    This AI-powered system automatically:
    - Generates quizzes from study text
    - Analyzes readability for grade alignment
    - Builds multilingual glossaries for inclusive learning
    ''')
    st.info('Use the left sidebar to explore modules like Analyze Text or Quizzes.')

elif page == 'Analyze Text':
    st.header('📘 Analyze Educational Text')
    text = st.text_area('Enter your educational text below:', height=250)
    uploaded = st.file_uploader('Or upload a .txt/.pdf file', type=['txt', 'pdf'])

    if uploaded and uploaded.type == 'application/pdf':
        from modules.pdf_utils import extract_text_from_pdf
        text = extract_text_from_pdf(uploaded)

    lang = st.selectbox('Target glossary language', ['hi', 'mr', 'en'])
    num_q = st.slider('Number of questions', 1, 15, 5)

    if st.button('Run Analysis'):
        if not text or not text.strip():
            st.warning('Please provide text to analyze.')
        else:
            with st.spinner('Processing...'):
                readability = ra.analyze_text(text)
                glossary = gb.build_glossary(text, target_lang=lang, top_k=15)
                quizzes = qg.generate(text, num_questions=num_q)
                sid = db.save_session(text, quizzes, readability, glossary)

            st.subheader('📈 Readability Metrics')
            st.json(readability)

            st.subheader('🌍 Multilingual Glossary')
            for word, entry in glossary.items():
                st.write(f"**{word}** → {entry['definition']} — {entry['translation']}")

            st.subheader('🧩 Generated Quiz Questions')
            for i, q in enumerate(quizzes, 1):
                st.write(f"**Q{i}.** {q['question']}")
                if q.get('choices'):
                    for idx, ch in enumerate(q['choices']):
                        st.write(f"   {chr(65 + idx)}. {ch}")

elif page == 'Quizzes':
    st.header('🗂 Saved Quiz Sessions')
    sessions = db.list_sessions()
    if not sessions:
        st.info('No previous sessions found.')
    else:
        for s in sessions:
            st.write(f"Session {s['id']} — {s['created_at']}")
            if st.button(f"Load Session {s['id']}"):
                data = db.load_session(s['id'])
                st.json(data)

elif page == 'Settings':
    st.header('⚙️ Settings')
    st.markdown('Adjust backend and model configurations here.')
    st.write('Set `OPENAI_API_KEY` in your environment if you want to use GPT-based question generation.')
```
